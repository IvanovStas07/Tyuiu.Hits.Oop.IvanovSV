<h2 style="color:DodgerBlue">Описание проекта:</h2>

----
Создать базовый класс CreditCard в C#, который будет представлять информацию
о кредитных картах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу CreditCard:
• Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок
действия (ExpiryDate).
• Методы:
o GetInfo(): метод для получения информации о кредитной карте в виде
строки.
o Pay(): метод для оплаты покупки с использованием карты.
o CheckBalance(): метод для проверки баланса на карте.

Требования к производным классам:
1. GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты,
такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть
переопределен для добавления информации о получении бонусных миль
при оплате покупки.
2. PremiumCard (PremiumCreditCard): Должен содержать дополнительные
атрибуты, такие как Ассистент поддержки (SupportAssistant).
Метод CheckBalance() должен быть переопределен для предоставления
возможности обратиться за помощью к ассистенту поддержки в случае
проблем с балансом.
3. CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Компания (Company).
Метод GetInfo() должен быть переопределен для включения информации о
компании в описании карты.

```
public class CreditCard
{
    public string CardNumber { get; protected set; }
    public string HolderName { get; protected set; }
    public DateTime ExpiryDate { get; protected set; }
    protected decimal Balance { get; set; }

    public CreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal initialBalance = 0)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = initialBalance;
    }

    // Метод для получения информации о карте
    public virtual string GetInfo()
    {
        return $"Карта: {CardNumber}, Владелец: {HolderName}, " +
               $"Срок действия: {ExpiryDate:MM/yyyy}, Баланс: {Balance:C}";
    }

    // Метод оплаты
    public virtual bool Pay(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма оплаты должна быть положительной.");
            return false;
        }

        if (Balance < amount)
        {
            Console.WriteLine($"Недостаточно средств на карте {CardNumber}. " +
                              $"Баланс: {Balance:C}, требуется: {amount:C}");
            return false;
        }

        Balance -= amount;
        Console.WriteLine($"Оплата {amount:C} с карты {CardNumber} выполнена успешно.");
        return true;
    }

    // Метод проверки баланса
    public virtual decimal CheckBalance()
    {
        Console.WriteLine($"Баланс карты {CardNumber}: {Balance:C}");
        return Balance;
    }
}

// Производный класс: GoldCreditCard
public class GoldCreditCard : CreditCard
{
    public int BonusMiles { get; private set; }

    public GoldCreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal initialBalance = 0)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        BonusMiles = 0;
    }

    // Переопределение Pay: начисление бонусных миль
    public override bool Pay(decimal amount)
    {
        bool success = base.Pay(amount);

        if (success)
        {
            int miles = (int)(amount / 100); // 1 миля за каждые 100 единиц валюты
            BonusMiles += miles;
            Console.WriteLine($"Начислено бонусных миль: {miles}. Всего миль: {BonusMiles}");
        }

        return success;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Бонусные мили: {BonusMiles}";
    }
}

// Производный класс: PremiumCreditCard
public class PremiumCreditCard : CreditCard
{
    public string SupportAssistant { get; private set; }

    public PremiumCreditCard(string cardNumber, string holderName, DateTime expiryDate,
                             decimal initialBalance, string supportAssistant)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        SupportAssistant = supportAssistant;
    }

    // Переопределение CheckBalance: помощь ассистента при проблемах с балансом
    public override decimal CheckBalance()
    {
        decimal balance = base.CheckBalance();

        if (balance < 1000)
        {
            Console.WriteLine($"Внимание! Низкий баланс. " +
                              $"Обратитесь к ассистенту поддержки: {SupportAssistant}");
        }
        else
        {
            Console.WriteLine($"При необходимости вы можете обратиться " +
                              $"к ассистенту поддержки: {SupportAssistant}");
        }

        return balance;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Ассистент поддержки: {SupportAssistant}";
    }
}

// Производный класс: CorporateCreditCard
public class CorporateCreditCard : CreditCard
{
    public string Company { get; private set; }

    public CorporateCreditCard(string cardNumber, string holderName, DateTime expiryDate,
                               decimal initialBalance, string company)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        Company = company;
    }

    // Переопределение GetInfo: добавление информации о компании
    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}";
    }
}

```

```
// Полиморфизм: список объектов базового типа CreditCard
List<CreditCard> cards = new List<CreditCard>
{
    new GoldCreditCard("1111 2222 3333 4444", "Иван Иванов",
                        new DateTime(2027, 12, 31), 50000),

    new PremiumCreditCard("5555 6666 7777 8888", "Петр Петров",
                            new DateTime(2026, 6, 30), 2000, "Анна Смирнова"),

    new CorporateCreditCard("9999 0000 1111 2222", "ООО Ромашка",
                            new DateTime(2028, 3, 31), 100000, "ООО Ромашка")
};

foreach (var card in cards)
{
    Console.WriteLine(card.GetInfo());
    card.CheckBalance();
    card.Pay(1500);
    card.CheckBalance();
    Console.WriteLine(new string('-', 50));
}
```
